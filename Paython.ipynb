```python
#Quetion 1
import pandas as pd
```

```python
Bank_Customer_Churn =  pd.read_csv(r'Bank_Customer_Churn_(U).csv')
```

```python
Bank_Customer_Churn.head(5)
```

```python
Bank_Customer_Churn["country"] = Bank_Customer_Churn["country"].fillna("India")
```

```python
# Quetion 2
female_count = (Bank_Customer_Churn["gender"] == "Female").sum()
```

```python
print(Bank_Customer_Churn["gender"].value_counts())
```

```python
Bank_Customer_Churn["gender"] = Bank_Customer_Churn["gender"].fillna(Bank_Customer_Churn["gender"].value_counts().idxmax())
```

```python
# Quetion 3
average_age = Bank_Customer_Churn["age"].mean()
```

```python
print("Average age:", average_age)
```

```python
Bank_Customer_Churn["age"] = Bank_Customer_Churn["age"].fillna(average_age)
```

```python
Bank_Customer_Churn["age"] = Bank_Customer_Churn["age"].astype(int)
```

```python
# Quetion 4
avg_balance = Bank_Customer_Churn["balance"].mean()
```

```python
# Maximum balance
max_balance = Bank_Customer_Churn["balance"].max()
```

```python
missing_idx = Bank_Customer_Churn[Bank_Customer_Churn["balance"].isna()].index
```

```python
# First 500 → average
Bank_Customer_Churn.loc[missing_idx[:500], "balance"] = avg_balance
```

```python
# Next 500 → maximum
Bank_Customer_Churn.loc[missing_idx[500:1000], "balance"] = max_balance
```

```python
# Remaining → 0
Bank_Customer_Churn.loc[missing_idx[1000:], "balance"] = 0
```

```python
# Exactly 2 decimal places
Bank_Customer_Churn["balance"] = Bank_Customer_Churn["balance"].round(2)
```

```python
# Quetion 5
missing_idx = Bank_Customer_Churn[Bank_Customer_Churn["active_member"].isna()].index
```

```python
half = len(missing_idx) // 2
```

```python
Bank_Customer_Churn.loc[missing_idx[:half], "active_member"] = 1
```

```python
Bank_Customer_Churn.loc[missing_idx[half:], "active_member"] = 0
```

```python
# Quetion 6
# Average salary
avg_salary = Bank_Customer_Churn["estimated_salary"].mean()
```

```python
# Fill missing values
Bank_Customer_Churn["estimated_salary"] = Bank_Customer_Churn["estimated_salary"].fillna(avg_salary)
```

```python
# Exactly 2 decimal places
Bank_Customer_Churn["estimated_salary"] = Bank_Customer_Churn["estimated_salary"].round(2)
```

```python
# Quetion 7
account_fact = Bank_Customer_Churn[['customer_id','credit_score','balance','products_number','credit_card','active_member','estimated_salary','churn']]
```

```python
account_fact.to_csv('Account_Fact(F).csv', index=False)
```
